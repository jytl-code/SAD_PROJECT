1. `commit` (The Local Save Point)
A commit is a permanent snapshot of your project at a specific moment in time.

When you run git commit -m "Added functions.py", Git takes a frozen snapshot of all your tracked files locally on your computer.

Important: Committing stays 100% on your Mac. It doesn’t send anything over the internet or to GitHub yet. It just means, "If I mess up my code tomorrow, I can always restore my project back to this exact working save point."

2. remote add origin (Adding GitHub's Address)
By default, Git on your Mac has no idea GitHub exists.

remote: Any version of your project hosted outside your computer (in the cloud or on another server).

origin: The default nickname Git uses for your primary cloud address.

When you ran git remote add origin [https://github.com/jytl-code/SAD_PROJECT.git](https://github.com/jytl-code/SAD_PROJECT.git), you were essentially adding GitHub to your Mac’s address book under the contact name "origin".

3. push (Uploading Your Save Points)
If a commit is creating a save file on your local machine, push is uploading those save files to the cloud.

When you run git push, your Mac looks up the nickname origin (your GitHub URL) and uploads all your recent local commit snapshots to GitHub.

This is what makes your code visible on github.com and accessible to your Linux machine.

4. origin/main (The Cloud Bookmark)
Git uses branch names to organize code.

main: Refers to the version of the code living locally on your Mac right now.

origin/main: Refers to the version of the main branch living up on GitHub (on origin).

When the terminal says branch 'main' set up to track 'origin/main', it is telling you: "Your local Mac branch is now linked directly to your GitHub branch." Next time you want to upload or download changes, you don't have to type out long URLs—Git knows automatically where to send and fetch the code.

Note:
you can use origin for every single project, and no, it will never confuse your Mac.

In fact, using origin as the nickname for every project is the universal standard in software development.

Why Git doesn't get confused
Git configuration is strictly isolated per project folder.

When you initialize a Git repository (git init), Git creates a hidden .git folder inside that specific project. Your remote address book lives inside that hidden folder (.git/config), not globally on your Mac.

Here is how your Mac sees it:

Inside /Users/you/SAD_PROJECT/:
The address book says: origin = [https://github.com/jytl-code/SAD_PROJECT.git](https://github.com/jytl-code/SAD_PROJECT.git)

Inside /Users/you/NEW_PROJECT/:
The address book says: origin = [https://github.com/jytl-code/NEW_PROJECT.git](https://github.com/jytl-code/NEW_PROJECT.git)

When you open a terminal, navigate into a folder, and run git push origin main, Git only checks the hidden .git folder of your current directory. It has no awareness of what origin means in any other project folder on your computer.
